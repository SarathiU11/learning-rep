###  Write a function that returns the cube of a number.


```python
def cube(n):
    return n * n * n
print(cube(5))
```

###  Write a function that returns the larger of two numbers. 

```python
def larg(a,b):
    if a>b:
        return a
    else:
        return b
print(larg(5,6))
```

### What are Default Arguments?

A default argument is a parameter that already has a value.

```python
def greet(name = "Guest"):
    print("hello",name)

greet()
greet("Sarathi")
```

### Keyword Arguments

Normally, arguments are matched by position

```python
def add(a,b):
    return a+b

print(add(5,4))
```

```python
def employee(name, salary):
    print(name, salary)

employee(salary=50000, name="Sarathi")
```

```python
def divide(a, b):
    return a / b
print(divide(b=5, a=20))
```
