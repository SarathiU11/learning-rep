````python
# Task 11

data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Print:
# 1. First row
print(data[0,1:0,2])# syntax : data:[ start:end ]
# 2. Second row
print(data[1 ,:])
# 3. First column
print(data[ : ,0])
# 4. Second column
print(data[: ,1])

# Write your code here
````
